```python
#LATIHAN 3
import pandas as pd

df = pd.read_csv(r'C:\Users\Reza Al Fath\Documents\EDA_python\data_kantin.csv')

print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Tugas Analisis 3: Dari hasil df.info(), kolom mana yang jumlah non-null-nya lebih sedikit dari
jumlah baris total? Apa artinya?

Kolom menu dan terjual. Karena bisa dilihat dari RangeIndex (total kolom), kolom menu hanya memiliki 107 non-null yang menandakan ada 3 kolom dengan null value. Sama juga dengan kolom terjual yang memiliki 6 kolom null value.

```python
#LATIHAN 4
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu'])
```

Tugas Analisis 4: Mengapa pada contoh di atas kolom terjual diisi (fillna) sedangkan baris
dengan menu kosong justru dihapus (dropna)? Diskusikan alasan logisnya.

Karena untuk kolom terjual, kita bisa menganggap 0 mewakili jika produk tersebut tidak laku/terdapat transaksi, tapi kita juga ingin memastikan agar kita masih bisa melakukan penghitungan hasil akhir penjualan. Itu sebabnya kita mengisi nilai null dengan 0.

Untuk menu kosong, kita mengambil keputusan untuk menghapusnya karena itu sama saja dengan kita tidak menjual menu itu. Jika kita membiarkan null ndi kolom menu, itu akan membuat nilai kolom lain terasa dirasa anonim (aneh jika kita mengetahui nilai harga atau terjual tanpa mengetahui apa nama menunnya)

```python
#LATIHAN 5
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```
